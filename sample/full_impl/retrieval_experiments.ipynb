```python
"""
test_reranking_standalone.py

Standalone check: run a Milvus search, then rerank the hits, and print
before/after so you can see what reranking actually changed.

Usage:
    python test_reranking_standalone.py
"""

import logging

from llama_index.core.schema import TextNode, NodeWithScore

from src.vector_storage.milvus_client import MilvusVectorClient
from src.retrieve.reranking import NodeReranker

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger("test_reranking")


def dicts_to_nodes(hits: list[dict]) -> list[NodeWithScore]:
    """
    Adapt MilvusVectorClient.search() output (list of plain dicts with
    'id', 'score', 'text', ...) into List[NodeWithScore], which is what
    NodeReranker.rerank() (and the rest of the LlamaIndex pipeline) expects.
    """
    nodes = []
    for hit in hits:
        node = TextNode(
            id_=str(hit.get("id", "")),
            text=hit.get("text", ""),
            metadata={k: v for k, v in hit.items() if k not in ("text", "id", "score")},
        )
        nodes.append(NodeWithScore(node=node, score=hit.get("score", 0.0)))
    return nodes


def print_ranked(label: str, nodes: list[NodeWithScore]):
    print(f"\n--- {label} ---")
    for i, n in enumerate(nodes, start=1):
        preview = n.node.get_content()[:120].replace("\n", " ")
        print(f"[{i}] score={n.score:.4f} | {preview}...")


if __name__ == "__main__":
    QUERY_TEXT = "What is Telangana govt start up initiative?"

    # 1. Get an embedding for the query.
    #    Swap this for whatever embed model your pipeline actually uses
    #    (must match the one used to build the collection).
    from llama_index.embeddings.huggingface import HuggingFaceEmbedding

    embed_model = HuggingFaceEmbedding(
        model_name="sentence-transformers/paraphrase-multilingual-mpnet-base-v2"
    )
    query_embedding = embed_model.get_query_embedding(QUERY_TEXT)

    # 2. Raw Milvus search (what you already ran).
    milvus_client = MilvusVectorClient(
        logger=logger,
        collection_name="funds_collection",
        dim=768,
        host="localhost",
        port="19530",
    )
    hits = milvus_client.search(query_embedding=query_embedding, top_k=8)
    print(f"Milvus returned {len(hits)} raw hits")

    # 3. Convert to NodeWithScore so the reranker can consume them.
    nodes_before = dicts_to_nodes(hits)
    print_ranked("BEFORE reranking (raw vector similarity order)", nodes_before)

    # 4. Rerank.
    reranker = NodeReranker(
        model_name="cross-encoder/ms-marco-MiniLM-L-6-v2",
        top_n=2,
    )
    nodes_after = reranker.rerank(nodes_before, QUERY_TEXT)
    print_ranked("AFTER reranking (cross-encoder relevance order)", nodes_after)
```
